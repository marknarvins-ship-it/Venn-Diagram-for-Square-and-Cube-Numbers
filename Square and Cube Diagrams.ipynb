```python
import matplotlib.pyplot as plt
from matplotlib_venn import venn2

# Define the sets
squares = {1, 4, 9, 16, 25, 36, 49, 64, 81, 100, 121, 144, 169, 196, 225,
           256, 289, 324, 361, 400, 441, 484, 529, 576, 625, 676, 729, 784, 841, 900, 961}
cubes = {1, 8, 27, 64, 125, 216, 343, 512, 729, 1000}

# Create Venn diagram with equal circle sizes
plt.figure(figsize=(12, 10))
v = venn2(subsets=(1, 1, 1), set_labels=('Square Numbers', 'Cube Numbers'))

# Set circle colors
v.get_patch_by_id('10').set_color('green')   # Squares only
v.get_patch_by_id('01').set_color('blue') # Cubes only
v.get_patch_by_id('11').set_color('cyan')  # Intersection

# Optional: adjust transparency
v.get_patch_by_id('10').set_alpha(0.6)
v.get_patch_by_id('01').set_alpha(0.6)
v.get_patch_by_id('11').set_alpha(0.8)

# Helper function to format numbers into multiple lines
def format_numbers(nums, max_per_line=5):
    nums = sorted(nums)
    lines = []
    for i in range(0, len(nums), max_per_line):
        lines.append(', '.join(map(str, nums[i:i+max_per_line])))
    return '\n'.join(lines)

# Add text labels with formatted values
v.get_label_by_id('10').set_text(format_numbers(squares - cubes))
v.get_label_by_id('01').set_text(format_numbers(cubes - squares))
v.get_label_by_id('11').set_text(format_numbers(squares & cubes))

# Reduce font size and set text color
for label_id in ['10', '01', '11']:
    v.get_label_by_id(label_id).set_fontsize(10)
    v.get_label_by_id(label_id).set_color('brown')

plt.title("Venn Diagram of Square and Cube Numbers (1–1000)")
plt.show()
```
